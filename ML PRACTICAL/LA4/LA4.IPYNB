```python
import numpy as np

# Function: y = (x + 3)^2
def f(x):
    return (x + 3) ** 2

# Derivative
def df(x):
    return 2 * (x + 3)

# Gradient Descent
x = 2
learning_rate = 0.1

for i in range(50):
    x = x - learning_rate * df(x)

print("Local minima at x =", round(x, 4))
print("Minimum value =", round(f(x), 4))
```
